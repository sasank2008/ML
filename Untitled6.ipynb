```python
import numpy as np
s=["Walking","Running","Resting"]#Hidden States
v={"Low":0,"Medium":1,"High":2}#Observation SYmbols 
#Transition Probability Matrix
A=np.array([
    [0.6,0.3,0.1],
    [0.2,0.5,0.3],
    [0.3,0.2,0.5]])
#Emission Probability Matrix
B=np.array([
    [0.6,0.3,0.1],
    [0.1,0.3,0.6],
    [0.7,0.2,0.1]])
#Initial Probability Vector
pi=np.array([0.5,0.3,0.2])
observations=["High","Medium","Low","High"]
obs=[v[o] for o in observations]
N=len(s)
T=len(obs)
alpha=np.zeros((T,N))
alpha[0]=pi*B[:,obs[0]]
for t in range(1,T):
    for j in range(N):
        alpha[t,j]=np.sum(alpha[t-1]*A[:,j])*B[j,obs[t]]
likelihood=np.sum(alpha[-1])
print(alpha)
print(likelihood)
```

```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM
```

```python
s=["Walking","Running","Resting"]#Hidden States
observation=["Low","Medium","High"]
obs_map={"Low":0,"Medium":1,"High":2}#Observation SYmbols 
#Transition Probability Matrix
A=np.array([
    [0.6,0.3,0.1],
    [0.2,0.5,0.3],
    [0.3,0.2,0.5]])
#Emission Probability Matrix
B=np.array([
    [0.6,0.3,0.1],
    [0.1,0.3,0.6],
    [0.7,0.2,0.1]])
#Initial Probability Vector
startprob=np.array([0.5,0.3,0.2])
observations=["High","Medium","Low","High"]
```

```python
model=CategoricalHMM(
    n_components = 3,
    init_params=""
)
model.startprob_ = startprob
model.transmat_ = A
model.emissionprob_ = B
```

```python
sequence = np.array([
    [obs_map["High"]],
    [obs_map["Medium"]],
    [obs_map["Low"]],
    [obs_map["High"]]
])
log_likelihood = model.score(sequence)
likelihood=np.exp(log_likelihood)
print("Log likelihood = ",log_likelihood)
print("Likelihood =",likelihood)
```

```python
test_sequences={
    "O1":["Low","Medium","High","Medium"],
    "O2":["High","High","High","High"],
    "O3":["Low","Low","Low","Low"],
    "O4":["Medium","Medium","Medium","Medium"]
}

for name,seq in test_sequences.items():
    obs=np.array([[obs_map[x]] for x in seq])
    log_prob = model.score(obs)
    prob=np.exp(log_prob)
    print(name)
    print(seq)
    print(log_prob)
    print(prob)
    print("-"*40)
```

```python
logprob,hidden_states=model.decode(sequence, algorithm="viterbi")
print(logprob)
for obs,state in zip(["High","Medium","Low","High"],hidden_states):
    print(f"{obs} --> {s[state]}")
```

```python
train_model = CategoricalHMM(n_components=3,n_iter=100,random_state=42)
train_model.fit(sequence)
```
