```python
import numpy as np

# Hidden states
states = ["Walking", "Running", "Resting"]

# Observation symbols
obs_symbols = ["Low", "Medium", "High"]

# Initial probabilities
pi = np.array([0.5, 0.3, 0.2])

# Transition matrix
A = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
])

# Emission matrix
B = np.array([
    [0.6, 0.3, 0.1],
    [0.1, 0.3, 0.6],
    [0.7, 0.2, 0.1]
])

# Observation sequence
observations = ["High", "Medium", "Low", "High"]

obs_index = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

T = len(observations)
N = len(states)

alpha = np.zeros((T, N))

# Initialization
alpha[0] = pi * B[:, obs_index[observations[0]]]

# Recursion
for t in range(1, T):
    for j in range(N):
        alpha[t, j] = np.sum(alpha[t-1] * A[:, j]) * B[j, obs_index[observations[t]]]

# Print Forward Table
print("Forward Probability Table (Alpha):\n")
print("Time\tObservation\tWalking\t\tRunning\t\tResting")

for t in range(T):
    print(f"{t+1}\t{observations[t]}\t\t"
          f"{alpha[t,0]:.6f}\t"
          f"{alpha[t,1]:.6f}\t"
          f"{alpha[t,2]:.6f}")

# Likelihood
likelihood = np.sum(alpha[-1])

print("\nLikelihood of Observation Sequence =", likelihood)

```

The **observation likelihood** computed using the **Forward Algorithm** and the likelihood obtained using **`hmmlearn`** represent the **same quantity**—the probability of observing a given sequence of observations under a specified Hidden Markov Model (HMM).

The Forward Algorithm calculates this likelihood by recursively computing the forward probabilities (α values) for each hidden state at every time step and summing the final probabilities:

Similarly, `hmmlearn` computes the likelihood internally using an optimized implementation of the Forward Algorithm. When you call methods such as `model.score(observations)`, it returns the **log-likelihood** of the observation sequence rather than the raw probability, because log values are numerically more stable for long sequences. If needed, the actual likelihood can be obtained by taking the exponential of the returned log-likelihood.

Therefore, if both the manually implemented Forward Algorithm and `hmmlearn` use the same initial probabilities, transition matrix, emission matrix, and observation sequence, they should produce equivalent results (allowing for small differences due to floating-point precision).
