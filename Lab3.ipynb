### The General Bayes' Theorem
At its foundation, the algorithm relies on Bayes' Theorem:

$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$

### The Naive Bayes Formula for Classification
When expanding this theorem for a machine learning dataset with multiple features, the formula looks like this:

$$P(y | x_1, x_2, \dots, x_n) = \frac{P(y) \prod_{i=1}^{n} P(x_i | y)}{P(x_1, x_2, \dots, x_n)}$$

Because the denominator (the evidence) is the same for all classes being compared, it is often written as a proportionality:

$$P(y | x_1, x_2, \dots, x_n) \propto P(y) \prod_{i=1}^{n} P(x_i | y)$$

---

### What Each Variable Denotes

*   **$y$**: The **Class Label** (e.g., Iris variety).
*   **$x_1, x_2, \dots, x_n$**: The **Features** (e.g., sepal length, petal width).
*   **$P(y | x_1, x_2, \dots, x_n)$**: The **Posterior Probability**. The probability that a data point belongs to class $y$, given its specific combination of features.
*   **$P(y)$**: The **Prior Probability**. The baseline probability of class $y$ appearing in the dataset.
*   **$P(x_i | y)$**: The **Likelihood**. The probability of observing feature $x_i$ given that the class is $y$. 
*   **$\prod$**: The **Product Operator**. It means you multiply all the individual likelihoods together.
*   **$P(x_1, x_2, \dots, x_n)$**: The **Evidence**. The overall probability of that exact combination of features appearing in the dataset.

## Calculate P(Y=y) for all possible y

```python
def calculate_prior(df,Y):
    classes = sorted(list(df[Y].unique()))
    prior = []
    for i in classes:
        prior.append(len(df[df[Y]==i])/len(df))
    return prior
 
```

```python
def calculate_likelihood_gaussian(df, feat_name, feat_value, Y, label):
    df = df[df[Y] == label]
    mean = df[feat_name].mean()
    std = df[feat_name].std()

    if std == 0:
        std = 1e-6  

    exponent = np.exp(-((feat_value - mean) ** 2) / (2 * std ** 2))
    return (1 / (np.sqrt(2 * np.pi) * std)) * exponent
```

```python
def naive_bayes_gaussian(df, X, Y):

    features = list(df.columns)[:-1]
    prior = calculate_prior(df, Y)
    labels = sorted(list(df[Y].unique()))

    Y_pred = []

    for x in X:

        likelihood = [1] * len(labels)

        for j in range(len(labels)):
            for i in range(len(features)):
                likelihood[j] *= calculate_likelihood_gaussian(
                    df,
                    features[i],
                    x[i],
                    Y,
                    labels[j]
                )

        post_prob = [likelihood[j] * prior[j] for j in range(len(labels))]

        # FIX HERE
        Y_pred.append(labels[np.argmax(post_prob)])

    return np.array(Y_pred)
```

```python
def confusion_matrix_scratch(y_true, y_pred):
    labels = sorted(list(set(y_true)))

    n = len(labels)
    matrix = [[0 for _ in range(n)] for _ in range(n)]

    label_to_index = {}
    for i, label in enumerate(labels):
        label_to_index[label] = i

    for actual, predicted in zip(y_true, y_pred):
        i = label_to_index[actual]
        j = label_to_index[predicted]
        matrix[i][j] += 1

    return labels, matrix
```

```python
def accuracy_score_scratch(y_true, y_pred):
    correct = 0

    for a, b in zip(y_true, y_pred):
        if a == b:
            correct += 1

    return correct / len(y_true)
```

```python
def macro_f1_score(y_true, y_pred):

    labels = sorted(list(set(y_true)))
    _, cm = confusion_matrix_scratch(y_true, y_pred)

    f1_scores = []

    for i in range(len(labels)):

        TP = cm[i][i]

        FP = 0
        for r in range(len(labels)):
            if r != i:
                FP += cm[r][i]

        FN = 0
        for c in range(len(labels)):
            if c != i:
                FN += cm[i][c]

        if TP + FP == 0:
            precision = 0
        else:
            precision = TP / (TP + FP)

        if TP + FN == 0:
            recall = 0
        else:
            recall = TP / (TP + FN)

        if precision + recall == 0:
            f1 = 0
        else:
            f1 = 2 * precision * recall / (precision + recall)

        f1_scores.append(f1)

    return sum(f1_scores) / len(f1_scores)
```

```python

data_shuffled = data.sample(frac=1, random_state=41).reset_index(drop=True)

split_index = int(0.8 * len(data_shuffled))

train = data_shuffled.iloc[:split_index]
test = data_shuffled.iloc[split_index:]

X_test = test.iloc[:, :-1].values
Y_test = test.iloc[:, -1].values


Y_pred = naive_bayes_gaussian(train, X_test, "variety")


labels, cm = confusion_matrix_scratch(Y_test, Y_pred)

print("Confusion Matrix")
for row in cm:
    print(row)

print("\nAccuracy:", accuracy_score_scratch(Y_test, Y_pred))
print("Macro F1 Score:", macro_f1_score(Y_test, Y_pred))
```
